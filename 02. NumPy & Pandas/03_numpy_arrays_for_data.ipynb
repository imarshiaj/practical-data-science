# NumPy Arrays for Data Work

Use arrays for fast numeric operations, masks, summaries, and missing numeric values.

> 🧭 This notebook focuses on practical data work. Examples are intentionally compact and do not cover machine learning.

## Learning goals
- Create arrays and inspect `shape`, `ndim`, and `dtype`.
- Use vectorized operations and boolean masks.
- Summarize data by axis and understand how NaN affects arithmetic.

NumPy arrays usually contain values of a compatible type. This makes them efficient for numerical work, but mixing unrelated types may coerce values or produce an object array.

```python
import numpy as np

values = np.array([12, 15, 18, 21])
print("values:", values)
print("shape:", values.shape, "ndim:", values.ndim, "dtype:", values.dtype)

matrix = np.array([[10, 20, 30], [40, 50, 60]])
print("matrix shape:", matrix.shape)
print("first row:", matrix[0])
print("second column:", matrix[:, 1])
```

## Vectorization and boolean masks
Vectorized expressions apply an operation to an entire array. They are often clearer and faster than manually looping over every value.

```python
prices = np.array([12.5, 9.0, 20.0, 7.5])
quantity = np.array([2, 4, 1, 5])
revenue = prices * quantity
print("Revenue per item:", revenue)
print("Total revenue:", revenue.sum())

expensive = prices >= 10
print("Expensive items:", prices[expensive])
print("Discounted prices:", np.where(prices >= 15, prices * 0.9, prices))
```

## Axis matters
For a 2-D array, `axis=0` reduces rows and returns a value for each column; `axis=1` reduces columns and returns a value for each row.

```python
sales = np.array([[10, 12, 9], [14, 8, 11], [7, 15, 13]])
print("Column totals:", sales.sum(axis=0))
print("Row totals:", sales.sum(axis=1))
print("Column means:", sales.mean(axis=0))
```

## Missing values and numeric stability
`np.nan` is a floating-point marker. Many ordinary operations propagate it; use `np.nanmean` or `np.isfinite` when that behavior matches your intent. Beware of integer arrays: inserting missing numeric values can change the dtype.

```python
measurements = np.array([2.4, np.nan, 3.1, 2.8, np.inf])
print("Missing mask:", np.isnan(measurements))
print("Finite mask:", np.isfinite(measurements))
print("Mean ignoring NaN (but not infinity):", np.nanmean(measurements))
print("Mean of finite values:", measurements[np.isfinite(measurements)].mean())

# Broadcasting: add one offset per column.
base = np.array([[1, 2, 3], [4, 5, 6]])
print(base + np.array([100, 200, 300]))
```

## Practice
Create an array of daily readings, select values above the mean, and calculate the mean while ignoring NaNs.

```python
readings = np.array([8.2, 9.1, np.nan, 7.8, 10.4])
mean_reading = np.nanmean(readings)
print("Mean:", round(mean_reading, 2))
print("Above mean:", readings[readings > mean_reading])
```

# Matplotlib for Data Analysis

Select, label, arrange, and save charts used to inspect data quality and patterns.

> 🧭 This notebook focuses on practical data work. Examples are intentionally compact and do not cover machine learning.

## Learning goals
- Choose a chart type based on the question and variable types.
- Label axes and units, add legends, and use subplots.
- Visualize distributions, comparisons, relationships, and trends.
- Save a figure with a predictable size and resolution.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

months = pd.date_range("2025-01-01", periods=6, freq="MS")
revenue = [120, 132, 128, 155, 162, 180]
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(months, revenue, marker="o", label="Revenue")
ax.set(title="Monthly revenue", xlabel="Month", ylabel="Revenue (currency units)")
ax.legend()
ax.grid(alpha=.25)
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

## Comparisons and distributions
- Bar charts compare a small set of categories.
- Histograms show numeric distributions; bin width can change the story.
- Boxplots summarize distributions but hide detail about shape and sample size.
- Scatterplots show relationships between two numeric variables.

```python
groups = ["North", "South", "East", "West"]
values = [95, 120, 82, 140]
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
axes[0].bar(groups, values)
axes[0].set_title("Category comparison")
axes[0].tick_params(axis="x", rotation=35)
axes[1].hist(np.random.default_rng(4).normal(50, 12, 300), bins=18, edgecolor="white")
axes[1].set_title("Distribution")
rng = np.random.default_rng(4)
axes[2].scatter(rng.uniform(0, 10, 60), rng.uniform(0, 100, 60), alpha=.65)
axes[2].set_title("Relationship")
axes[2].set_xlabel("X")
axes[2].set_ylabel("Y")
plt.tight_layout()
plt.show()
```

## Scale, annotation, and saving
A truncated y-axis can exaggerate differences in bar charts. Log scales can help with orders of magnitude, but label them clearly and do not apply them blindly to values that can be zero or negative.

```python
x = np.arange(4)
counts = [100, 105, 110, 115]
fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(["Q1", "Q2", "Q3", "Q4"], counts)
ax.bar_label(bars, fmt="%d")
ax.set_ylim(0, 130)
ax.set_ylabel("Count")
ax.set_title("A clearly labeled bar chart with a zero baseline")
plt.tight_layout()
# In a real report: fig.savefig("quarterly_counts.png", dpi=160, bbox_inches="tight")
plt.show()
```

# Interactive Charts with Plotly Express

Build interactive charts for exploratory inspection, filtering, and hover details.

> 🧭 This notebook focuses on practical data work. Examples are intentionally compact and do not cover machine learning.

## Learning goals
- Create interactive scatter, line, and bar charts with Plotly Express.
- Add hover metadata and facet panels.
- Use interactive charts for exploration while preserving a static export option.

Plotly charts support hover, zoom, and legend toggling in notebook environments. Some export options require extra dependencies such as Kaleido; interactive notebook output does not.

```python
import pandas as pd
import numpy as np
import plotly.express as px

rng = np.random.default_rng(17)
df = pd.DataFrame({
    "month": pd.date_range("2025-01-01", periods=36, freq="MS"),
    "region": np.tile(["North", "South", "West"], 12),
    "revenue": rng.integers(70, 180, 36),
    "orders": rng.integers(10, 60, 36),
})
df.head()
```

```python
fig = px.scatter(
    df, x="orders", y="revenue", color="region", size="orders",
    hover_data={"month": True}, title="Revenue versus order count"
)
fig.show()
```

## Lines, categories, and facets
For time-series lines, sort data by time first. A line connecting unordered categories can imply a sequence that does not exist.

```python
monthly_region = df.sort_values("month")
fig = px.line(monthly_region, x="month", y="revenue", color="region", markers=True,
              title="Monthly revenue by region")
fig.show()

fig = px.histogram(df, x="revenue", color="region", barmode="overlay", opacity=.65,
                   title="Revenue distribution by region")
fig.show()
```

## Interactive charts: good habits
- Include units and useful hover fields.
- Do not hide the scale or omit missing groups without explanation.
- Interactive features should answer a clear analysis question, not merely decorate a report.
- If exporting a static image fails, check whether the optional renderer dependency is installed.

```python
region_summary = df.groupby("region", as_index=False).agg(
    total_revenue=("revenue", "sum"), average_order_count=("orders", "mean")
)
fig = px.bar(region_summary, x="region", y="total_revenue", text_auto=True,
             title="Total revenue by region", labels={"total_revenue": "Revenue (units)"})
fig.show()
display(region_summary)
```

## Practice
Build an interactive bar chart from a grouped DataFrame. Use an informative title and labels, and add a column to hover data.

```python
fig = px.bar(region_summary, x="region", y="total_revenue", hover_data=["average_order_count"],
             title="Practice: revenue by region", labels={"total_revenue": "Revenue (units)"})
fig.show()
```

## Learning goals
- Recognize common Python data types and containers.
- Use indexing, loops, comprehensions, and simple conditions.
- Write small, readable snippets that help with data tasks.

Python is the glue around data libraries. You do not need advanced software engineering before you can start analyzing data, but you should be comfortable reading values, handling containers, and checking assumptions.

```python
# Basic scalar types
record_count = 128
average_rating = 4.35
source_name = "survey"
is_verified = True
missing_value = None

for name, value in {
    "record_count": record_count,
    "average_rating": average_rating,
    "source_name": source_name,
    "is_verified": is_verified,
    "missing_value": missing_value,
}.items():
    print(f"{name:16} -> {value!r:12} ({type(value).__name__})")
```

## Lists and dictionaries
A list stores an ordered sequence. A dictionary maps keys to values. These containers are useful for small configuration objects, API responses, and records before converting them into a DataFrame.

```python
cities = ["Baku", "Tbilisi", "Yerevan"]
print(cities[0])
print(cities[-1])
print(cities[1:])

site = {"name": "Central Library", "visits": 240, "active": True}
print(site["name"])
print(site.get("rating", "not provided"))

# Update a value and add a field.
site["visits"] += 10
site["rating"] = 4.7
print(site)
```

## Loops, conditions, and comprehensions
Use a loop when each item needs its own processing step. A comprehension is compact when the transformation is simple.

```python
measurements = [12, 15, None, 9, 18, None]
valid = []
for value in measurements:
    if value is not None:
        valid.append(value)
print("Valid values:", valid)

# Equivalent compact form
valid_again = [value for value in measurements if value is not None]
print("Same result:", valid_again)

labels = ["low" if value < 12 else "high" for value in valid]
print(labels)
```

## Useful built-ins and common pitfalls
- `len`, `sum`, `min`, `max`, and `sorted` are useful for quick checks.
- `sorted(values)` returns a new list; `values.sort()` changes the list in place and returns `None`.
- `and` / `or` combine conditions; use parentheses to make complicated conditions readable.
- `None` is checked with `is None`, not `== None`.

```python
values = [7, 2, 11, 2, 5]
print("Count:", len(values), "Total:", sum(values), "Range:", max(values) - min(values))
print("Sorted copy:", sorted(values))
values.sort()
print("Original after .sort():", values)

value = None
print("Missing?", value is None)

scores = [48, 76, 91, 63]
passed = [score for score in scores if score >= 60]
print("Passed:", passed)
```
